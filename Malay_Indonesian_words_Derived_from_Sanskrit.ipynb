```python
!pip -q install pandas requests beautifulsoup4 tabulate

import pandas as pd
import requests
from bs4 import BeautifulSoup
from tabulate import tabulate

url = "https://en.wikipedia.org/wiki/List_of_Indic_loanwords_in_Indonesian"

html = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
).text

soup = BeautifulSoup(html, "html.parser")

data = []

for table in soup.find_all("table"):

    rows = table.find_all("tr")

    for row in rows[1:]:

        cells = row.find_all(["td", "th"])

        if len(cells) >= 5:

            values = [
                cell.get_text(" ", strip=True)
                for cell in cells
            ]

            indonesian = values[0]

            sanskrit_script = values[2]

            sanskrit_translit = values[3]

            english_meaning = values[4]

            if (
                indonesian
                and sanskrit_script
                and sanskrit_translit
                and english_meaning
            ):

                sanskrit = (
                    f"{sanskrit_translit} "
                    f"({sanskrit_script})"
                )

                data.append([
                    indonesian,
                    sanskrit,
                    english_meaning
                ])

df = pd.DataFrame(
    data,
    columns=[
        "Malay/Indonesian",
        "Sanskrit",
        "English Meaning"
    ]
)

df = df.drop_duplicates()

print(
    tabulate(
        df.head(200),
        headers="keys",
        tablefmt="grid",
        showindex=False
    )
)

display(df)

df.to_csv(
    "Malay_Indonesian_Sanskrit_Words.csv",
    index=False
)

print(f"\nTotal Words: {len(df)}")
```
